```python
#cropped e aligned
import os
import unicodedata
import cv2
from deepface import DeepFace
# per il dataset originale
#input_folder = r"C:\Users\Asus\Desktop\progetto_caravaggio\dataset\originali\caravaggio"
#output_folder = r"C:\Users\Asus\Desktop\progetto_caravaggio\dataset\pre_processato\colorati\normali"

# per il dataset generative_ia
input_folder = r"C:\Users\Asus\Desktop\progetto_caravaggio\dataset\generative_ia\dataset_caravaggio"
output_folder = r"C:\Users\Asus\Desktop\progetto_caravaggio\dataset\pre_processato\colorati\generative_ia"

os.makedirs(output_folder, exist_ok=True)

valid_extensions = ('.jpg', '.jpeg', '.png', '.tif', '.tiff')

for img_name in os.listdir(input_folder):
    img_path = os.path.join(input_folder, img_name)
    
    if os.path.isfile(img_path) and img_name.lower().endswith(valid_extensions):
        try:
            print(f"Rilevamento, ritaglio e allineamento per: {img_name}...")
            
            # 1. Detection + Crop + Alignment in un unico passaggio
            faces = DeepFace.extract_faces(
                img_path=img_path,
                detector_backend='retinaface',
                enforce_detection=False,
                align=True
            )

            # 2. Pulizia del nome file per evitare problemi con caratteri non-ASCII
            clean_name = unicodedata.normalize('NFKD', img_name).encode('ASCII', 'ignore').decode('ASCII')
            base_name = os.path.splitext(clean_name)[0]

            # 3. Conversione e salvataggio di ogni volto estratto
            for i, face_obj in enumerate(faces):
                face_img = face_obj["face"]
                
                # Conversione da float32 [0, 1] RGB a uint8 [0, 255] BGR per OpenCV
                face_img_rescaled = (face_img * 255).astype('uint8')
                face_img_bgr = cv2.cvtColor(face_img_rescaled, cv2.COLOR_RGB2BGR)

                save_path = os.path.join(output_folder, f"{base_name}_face{i}.jpg")
                cv2.imwrite(save_path, face_img_bgr)
                print(f"   -> Salvato: {save_path}")

        except Exception as e:
            print(f"❌ Errore su {img_name}: {e}")

print("\n🎉 Crop e allineamento completati in un unico passaggio!")
```

```python
# per i bianco e enro cambiare path base:input e dataset_folder
# 1. Cartella radice dei volti ritagliati a colori
base_input = r"C:\Users\Asus\Desktop\progetto_caravaggio\dataset\pre_processato\colorati"

# 2. Cartelle di destinazione per lo smistamento di genere
dataset_folder = r"C:\Users\Asus\Desktop\progetto_caravaggio\dataset\genere\colorati"
male_folder = os.path.join(dataset_folder, "Male")
female_folder = os.path.join(dataset_folder, "Female")

os.makedirs(male_folder, exist_ok=True)
os.makedirs(female_folder, exist_ok=True)

valid_extensions = ('.jpg', '.jpeg', '.png', '.tif', '.tiff')

if not os.path.exists(base_input):
    raise FileNotFoundError(f"Directory principale non trovata: {base_input}")

# Rileva automaticamente tutte le sottocartelle esistenti (es. 'generative_ia', 'caravaggio', ecc.)
sottocartelle = [d for d in os.listdir(base_input) if os.path.isdir(os.path.join(base_input, d))]

print(f"Sottocartelle rilevate in {base_input}: {sottocartelle}\n")

for sorgente in sottocartelle:
    cartella_corrente = os.path.join(base_input, sorgente)
    print(f"\n📂 Elaborazione sottocartella: {sorgente}")
    
    for img_name in os.listdir(cartella_corrente):
        img_path = os.path.join(cartella_corrente, img_name)
        
        if os.path.isfile(img_path) and img_name.lower().endswith(valid_extensions):
            try:
                # Pulizia dei caratteri speciali/accenti per compatibilità con OpenCV e DeepFace
                clean_name = unicodedata.normalize('NFKD', img_name).encode('ASCII', 'ignore').decode('ASCII')
                dest_filename = f"{sorgente}_{clean_name}"

                # Classificazione genere (senza ri-eseguire detection)
                result = DeepFace.analyze(
                    img_path=img_path,
                    actions=['gender'],
                    enforce_detection=False,
                    detector_backend='skip'
                )

                if isinstance(result, list):
                    result = result[0]

                gender = result['dominant_gender']
                print(f" -> [{sorgente}] {img_name} -> {gender}")

                # Smistamento
                if gender == 'Man':
                    dest_path = os.path.join(male_folder, dest_filename)
                else:
                    dest_path = os.path.join(female_folder, dest_filename)
                
                shutil.copy2(img_path, dest_path)

            except Exception as e:
                print(f"❌ Errore su {img_name}: {e}")

print("\n🎉 Processo completato: tutti i volti sono stati smistati con successo!")
```
